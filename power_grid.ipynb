```python

# ================================================================
# POWER GRID DATASET - COMPLETE 11 ML MODELS IMPLEMENTATION
# GOOGLE COLAB - SINGLE CELL
# ================================================================

!pip -q install kagglehub scikit-learn pandas numpy matplotlib seaborn

import os
import glob
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import kagglehub

from sklearn.model_selection import train_test_split, learning_curve
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import (
    StandardScaler, OneHotEncoder, LabelEncoder,
    label_binarize
)
from sklearn.impute import SimpleImputer

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, confusion_matrix, classification_report,
    roc_curve, auc, roc_auc_score,
    mean_absolute_error, mean_squared_error, r2_score
)

from sklearn.linear_model import LogisticRegression, LinearRegression, Ridge
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.ensemble import (
    RandomForestClassifier, ExtraTreesClassifier,
    GradientBoostingClassifier, AdaBoostClassifier,
    BaggingClassifier, RandomForestRegressor,
    ExtraTreesRegressor, GradientBoostingRegressor,
    AdaBoostRegressor, BaggingRegressor,
    HistGradientBoostingRegressor, RandomForestRegressor
)
from sklearn.svm import SVC, SVR
from sklearn.neighbors import KNeighborsClassifier, KNeighborsRegressor
from sklearn.naive_bayes import GaussianNB
from sklearn.neural_network import MLPClassifier, MLPRegressor
from sklearn.ensemble import HistGradientBoostingRegressor

# ------------------------------------------------
# 1. DOWNLOAD DATASET
# ------------------------------------------------

print("="*65)
print("POWER GRID DATASET - 11 MACHINE LEARNING MODELS")
print("="*65)

path = kagglehub.dataset_download("ziya07/power-grid")

print("\nDataset downloaded:", path)

csv_files = glob.glob(
    os.path.join(path, "**", "*.csv"),
    recursive=True
)

if not csv_files:
    raise FileNotFoundError(
        "No CSV file found in the downloaded dataset."
    )

# Load the largest CSV file
csv_file = max(csv_files, key=os.path.getsize)
df = pd.read_csv(csv_file)

print("\nSelected file:", os.path.basename(csv_file))
print("Dataset shape:", df.shape)

# ------------------------------------------------
# 2. DATASET ANALYSIS
# ------------------------------------------------

print("\nFIRST FIVE ROWS")
display(df.head())

print("\nCOLUMN NAMES")
print(df.columns.tolist())

print("\nDATA TYPES")
print(df.dtypes)

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nDUPLICATE RECORDS")
print(df.duplicated().sum())

print("\nSTATISTICAL DESCRIPTION")
display(df.describe(include="all").T)

# Remove duplicate records
df = df.drop_duplicates().reset_index(drop=True)

# Remove completely empty columns
df = df.dropna(axis=1, how="all")

# Clean column names
df.columns = df.columns.astype(str).str.strip()

# ------------------------------------------------
# 3. INPUT AND OUTPUT VARIABLES
# ------------------------------------------------

# If automatic target selection is incorrect,
# enter your actual target column name here.
TARGET_COLUMN = None

# Common target column names
target_names = [
    "label", "target", "class", "fault",
    "fault_type", "fault_class", "status",
    "condition", "anomaly", "output",
    "efficiency_class", "grid_status",
    "power_loss", "efficiency"
]

if TARGET_COLUMN is None:
    lower_columns = {
        str(c).lower().strip(): c for c in df.columns
    }

    for name in target_names:
        if name in lower_columns:
            TARGET_COLUMN = lower_columns[name]
            break

# Fallback: use the last column, but verify it
# against the actual dataset documentation.
if TARGET_COLUMN is None:
    TARGET_COLUMN = df.columns[-1]
    print("\nWARNING: No recognized target name found.")
    print("Using the last column temporarily.")
    print("Verify TARGET_COLUMN before using the results.")

if TARGET_COLUMN not in df.columns:
    raise ValueError("Target column does not exist!")

print("\nOUTPUT VARIABLE (Y):", TARGET_COLUMN)

# Remove rows where target is missing
df = df.dropna(subset=[TARGET_COLUMN]).copy()

# Input variables
X = df.drop(columns=[TARGET_COLUMN])
y = df[TARGET_COLUMN]

# Remove columns containing no useful information
X = X.loc[:, X.nunique(dropna=False) > 1]

print("\nINPUT VARIABLES (X):")
print(X.columns.tolist())

print("\nOUTPUT VARIABLE (Y):")
print(y.name)

print("\nNUMBER OF INPUT FEATURES:", X.shape[1])
print("NUMBER OF OUTPUT RECORDS:", len(y))

# ------------------------------------------------
# 4. AUTOMATIC CLASSIFICATION / REGRESSION DETECTION
# ------------------------------------------------

is_classification = (
    y.dtype == "object"
    or str(y.dtype) == "category"
    or y.dtype == "bool"
    or y.nunique() <= 20
)

if is_classification:
    print("\nPROBLEM TYPE: CLASSIFICATION")

    # Convert target into numerical class labels
    encoder = LabelEncoder()
    y = encoder.fit_transform(y.astype(str))

    print("Classes:", encoder.classes_)
    print("Number of classes:", len(encoder.classes_))

    if len(np.unique(y)) < 2:
        raise ValueError("Target must contain at least 2 classes.")

    if pd.Series(y).value_counts().min() < 2:
        raise ValueError(
            "Each class needs at least two samples for a stratified split."
        )

else:
    print("\nPROBLEM TYPE: REGRESSION")
    y = pd.to_numeric(y, errors="coerce")
    valid = y.notna()
    X = X.loc[valid]
    y = y.loc[valid].astype(float)

# ------------------------------------------------
# 5. PREPROCESSING
# ------------------------------------------------

numeric_features = X.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features = X.select_dtypes(
    exclude=["number"]
).columns.tolist()

try:
    onehot = OneHotEncoder(
        handle_unknown="ignore",
        max_categories=20,
        sparse_output=False
    )
except TypeError:
    onehot = OneHotEncoder(
        handle_unknown="ignore",
        max_categories=20,
        sparse=False
    )

numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", onehot)
])

transformers = []

if numeric_features:
    transformers.append(
        ("numeric", numeric_transformer, numeric_features)
    )

if categorical_features:
    transformers.append(
        ("categorical", categorical_transformer,
         categorical_features)
    )

preprocessor = ColumnTransformer(
    transformers=transformers,
    remainder="drop"
)

# ------------------------------------------------
# 6. TRAINING AND TESTING SPLIT
# ------------------------------------------------

if len(X) < 10:
    raise ValueError("Dataset is too small for model training.")

if is_classification:
    X_train, X_test, y_train, y_test = train_test_split(
        X, y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )
else:
    X_train, X_test, y_train, y_test = train_test_split(
        X, y,
        test_size=0.20,
        random_state=42
    )

print("\nTRAINING RECORDS:", len(X_train))
print("TESTING RECORDS:", len(X_test))

# ------------------------------------------------
# 7. DEFINE 11 ML MODELS
# ------------------------------------------------

if is_classification:

    models = {
        "Logistic Regression": LogisticRegression(
            max_iter=1000
        ),

        "Decision Tree": DecisionTreeClassifier(
            random_state=42
        ),

        "Random Forest": RandomForestClassifier(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        ),

        "Extra Trees": ExtraTreesClassifier(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        ),

        "Gradient Boosting": GradientBoostingClassifier(
            random_state=42
        ),

        "AdaBoost": AdaBoostClassifier(
            n_estimators=50,
            random_state=42
        ),

        "SVM": SVC(
            probability=True,
            random_state=42
        ),

        "KNN": KNeighborsClassifier(
            n_neighbors=min(5, len(X_train))
        ),

        "Naive Bayes": GaussianNB(),

        "Neural Network (MLP)": MLPClassifier(
            hidden_layer_sizes=(64, 32),
            max_iter=250,
            early_stopping=True,
            random_state=42
        ),

        "Bagging": BaggingClassifier(
            n_estimators=50,
            random_state=42,
            n_jobs=-1
        )
    }

else:

    models = {
        "Linear Regression": LinearRegression(),

        "Ridge Regression": Ridge(),

        "Decision Tree": DecisionTreeRegressor(
            random_state=42
        ),

        "Random Forest": RandomForestRegressor(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        ),

        "Extra Trees": ExtraTreesRegressor(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        ),

        "Gradient Boosting": GradientBoostingRegressor(
            random_state=42
        ),

        "AdaBoost": AdaBoostRegressor(
            n_estimators=50,
            random_state=42
        ),

        "SVR": SVR(),

        "KNN": KNeighborsRegressor(
            n_neighbors=min(5, len(X_train))
        ),

        "Neural Network (MLP)": MLPRegressor(
            hidden_layer_sizes=(64, 32),
            max_iter=250,
            early_stopping=True,
            random_state=42
        ),

        "Hist Gradient Boosting": HistGradientBoostingRegressor(
            random_state=42
        )
    }

print("\nTOTAL MODELS:", len(models))

# ------------------------------------------------
# 8. TRAIN ALL 11 MODELS
# ------------------------------------------------

results = []
trained_models = {}
predictions = {}

print("\n" + "="*65)
print("MODEL TRAINING STARTED")
print("="*65)

for name, model in models.items():

    print("\nTraining:", name)

    pipeline = Pipeline([
        ("preprocessing", preprocessor),
        ("model", model)
    ])

    try:
        pipeline.fit(X_train, y_train)

        pred = pipeline.predict(X_test)

        trained_models[name] = pipeline
        predictions[name] = pred

        if is_classification:

            accuracy = accuracy_score(y_test, pred)

            precision = precision_score(
                y_test, pred,
                average="weighted",
                zero_division=0
            )

            recall = recall_score(
                y_test, pred,
                average="weighted",
                zero_division=0
            )

            f1 = f1_score(
                y_test, pred,
                average="weighted",
                zero_division=0
            )

            try:
                if len(np.unique(y_test)) == 2:
                    auc_score = roc_auc_score(
                        y_test,
                        pipeline.predict_proba(X_test)[:, 1]
                    )
                else:
                    auc_score = roc_auc_score(
                        y_test,
                        pipeline.predict_proba(X_test),
                        multi_class="ovr",
                        average="weighted"
                    )
            except:
                auc_score = np.nan

            results.append({
                "Model": name,
                "Accuracy (%)": accuracy * 100,
                "Precision (%)": precision * 100,
                "Recall (%)": recall * 100,
                "F1 Score (%)": f1 * 100,
                "ROC-AUC": auc_score
            })

            print("Accuracy:", round(accuracy * 100, 2), "%")

        else:

            mae = mean_absolute_error(y_test, pred)
            rmse = np.sqrt(mean_squared_error(y_test, pred))
            r2 = r2_score(y_test, pred)

            results.append({
                "Model": name,
                "MAE": mae,
                "RMSE": rmse,
                "R2 Score (%)": r2 * 100
            })

            print("R2 Score:", round(r2 * 100, 2), "%")

    except Exception as e:
        print("MODEL FAILED:", name)
        print("Reason:", str(e)[:500])

if not results:
    raise RuntimeError("All models failed. Check dataset and target.")

results_df = pd.DataFrame(results)

# ------------------------------------------------
# 9. MODEL COMPARISON
# ------------------------------------------------

print("\n" + "="*65)
print("FINAL MODEL COMPARISON")
print("="*65)

if is_classification:
    results_df = results_df.sort_values(
        "Accuracy (%)",
        ascending=False
    )
else:
    results_df = results_df.sort_values(
        "R2 Score (%)",
        ascending=False
    )

display(results_df.round(4))

results_df.to_csv(
    "Power_Grid_11_Model_Results.csv",
    index=False
)

best_name = results_df.iloc[0]["Model"]
best_model = trained_models[best_name]

print("\nBEST MODEL:", best_name)

if is_classification:
    best_accuracy = results_df.iloc[0]["Accuracy (%)"]
    print("BEST ACCURACY:", round(best_accuracy, 2), "%")

    if best_accuracy >= 85:
        print("Target accuracy of 85% achieved.")
    else:
        print("85% target not achieved. Further tuning may help.")
else:
    print("Best R2:", results_df.iloc[0]["R2 Score (%)"], "%")
    print("Note: R2 is not classification accuracy.")

# ------------------------------------------------
# 10. ACCURACY / PERFORMANCE GRAPH
# ------------------------------------------------

plt.figure(figsize=(14, 6))

if is_classification:
    sns.barplot(
        data=results_df,
        x="Model",
        y="Accuracy (%)"
    )
    plt.ylabel("Accuracy (%)")
    plt.title("Accuracy Comparison of 11 ML Models")
else:
    sns.barplot(
        data=results_df,
        x="Model",
        y="R2 Score (%)"
    )
    plt.ylabel("R2 Score (%)")
    plt.title("R2 Comparison of 11 Regression Models")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# ------------------------------------------------
# 11. PRECISION, RECALL AND F1 SCORE
# ------------------------------------------------

if is_classification:

    results_df.set_index("Model")[
        ["Precision (%)", "Recall (%)", "F1 Score (%)"]
    ].plot(kind="bar", figsize=(14, 6))

    plt.title("Precision, Recall and F1 Score Comparison")
    plt.ylabel("Score (%)")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

# ------------------------------------------------
# 12. ROC CURVES
# ------------------------------------------------

if is_classification:

    plt.figure(figsize=(10, 7))

    for name, pipeline in trained_models.items():

        try:
            probabilities = pipeline.predict_proba(X_test)

            if len(np.unique(y)) == 2:
                fpr, tpr, _ = roc_curve(
                    y_test, probabilities[:, 1]
                )

                roc_value = auc(fpr, tpr)

                plt.plot(
                    fpr, tpr,
                    label=f"{name} (AUC={roc_value:.3f})"
                )

            else:
                y_bin = label_binarize(
                    y_test,
                    classes=np.arange(len(encoder.classes_))
                )

                fpr, tpr, _ = roc_curve(
                    y_bin.ravel(),
                    probabilities.ravel()
                )

                roc_value = auc(fpr, tpr)

                plt.plot(
                    fpr, tpr,
                    label=f"{name} (Micro AUC={roc_value:.3f})"
                )

        except Exception as e:
            print("ROC unavailable for:", name)

    plt.plot([0, 1], [0, 1], linestyle="--")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title("ROC Curve Comparison - 11 ML Models")
    plt.legend(loc="lower right", fontsize=8)
    plt.grid()
    plt.tight_layout()
    plt.show()

# ------------------------------------------------
# 13. CONFUSION MATRIX
# ------------------------------------------------

if is_classification:

    best_predictions = predictions[best_name]

    cm = confusion_matrix(
        y_test,
        best_predictions
    )

    plt.figure(figsize=(8, 6))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues"
    )

    plt.title("Confusion Matrix - " + best_name)
    plt.xlabel("Predicted Class")
    plt.ylabel("Actual Class")
    plt.tight_layout()
    plt.show()

    print("\nCLASSIFICATION REPORT")

    print(classification_report(
        y_test,
        best_predictions,
        target_names=encoder.classes_.astype(str),
        zero_division=0
    ))

# ------------------------------------------------
# 14. TRAINING AND VALIDATION GRAPHS
# ------------------------------------------------

# Learning curves estimate performance at different
# training-set sizes using cross-validation.
# Limit sample size to keep Colab execution manageable.

print("\nGenerating training and validation curves...")

learning_models = list(trained_models.items())

if len(X_train) > 5000:
    if is_classification:
        lc_X, _, lc_y, _ = train_test_split(
            X_train, y_train,
            train_size=5000,
            random_state=42,
            stratify=y_train
        )
    else:
        lc_X, _, lc_y, _ = train_test_split(
            X_train, y_train,
            train_size=5000,
            random_state=42
        )
else:
    lc_X, lc_y = X_train, y_train

for name, pipeline in learning_models:

    print("Learning curve:", name)

    try:
        cv_count = 3

        if is_classification:
            min_class_count = pd.Series(lc_y).value_counts().min()
            cv_count = min(3, int(min_class_count))

        if cv_count < 2:
            print("Not enough samples for learning curve.")
            continue

        train_sizes, train_scores, validation_scores = learning_curve(
            pipeline,
            lc_X,
            lc_y,
            cv=cv_count,
            train_sizes=np.linspace(0.3, 1.0, 4),
            scoring="accuracy" if is_classification
                    else "r2",
            n_jobs=-1,
            shuffle=True,
            random_state=42
        )

        plt.figure(figsize=(8, 5))

        plt.plot(
            train_sizes,
            train_scores.mean(axis=1),
            marker="o",
            label="Training Score"
        )

        plt.plot(
            train_sizes,
            validation_scores.mean(axis=1),
            marker="o",
            label="Validation Score"
        )

        plt.xlabel("Training Samples")
        plt.ylabel(
            "Accuracy" if is_classification else "R2 Score"
        )

        plt.title(name + " - Training vs Validation")
        plt.legend()
        plt.grid()
        plt.tight_layout()
        plt.show()

    except Exception as e:
        print("Learning curve unavailable:", str(e)[:250])

# ------------------------------------------------
# 15. LOSS / OPTIMIZATION GRAPH
# ------------------------------------------------

print("\nGenerating available loss and optimization graphs...")

for name, pipeline in trained_models.items():

    fitted_model = pipeline.named_steps["model"]

    # Neural network optimization loss
    if hasattr(fitted_model, "loss_curve_"):

        plt.figure(figsize=(8, 5))

        plt.plot(
            fitted_model.loss_curve_,
            label="Training Loss"
        )

        plt.xlabel("Iteration")
        plt.ylabel("Loss")
        plt.title(name + " - Optimization / Loss Curve")
        plt.legend()
        plt.grid()
        plt.tight_layout()
        plt.show()

    # Gradient boosting staged loss
    elif hasattr(fitted_model, "train_score_"):

        plt.figure(figsize=(8, 5))

        plt.plot(
            fitted_model.train_score_,
            label="Training Loss"
        )

        plt.xlabel("Boosting Iteration")
        plt.ylabel("Loss")
        plt.title(name + " - Optimization Curve")
        plt.legend()
        plt.grid()
        plt.tight_layout()
        plt.show()

# ------------------------------------------------
# 16. FEATURE IMPORTANCE
# ------------------------------------------------

print("\nGenerating feature importance...")

try:

    fitted = best_model.named_steps["model"]

    if hasattr(fitted, "feature_importances_"):

        transformed_names = (
            best_model.named_steps["preprocessing"]
            .get_feature_names_out()
        )

        importance = fitted.feature_importances_

        feature_df = pd.DataFrame({
            "Feature": transformed_names,
            "Importance": importance
        }).sort_values(
            "Importance",
            ascending=False
        ).head(15)

        plt.figure(figsize=(10, 6))

        sns.barplot(
            data=feature_df,
            x="Importance",
            y="Feature"
        )

        plt.title("Top 15 Important Features")
        plt.tight_layout()
        plt.show()

        display(feature_df)

    else:
        print("Feature importance not available for this model.")

except Exception as e:
    print("Feature importance unavailable:", str(e)[:250])

# ------------------------------------------------
# 17. ACTUAL VS PREDICTED GRAPH
# ------------------------------------------------

plt.figure(figsize=(8, 6))

if is_classification:

    comparison = pd.DataFrame({
        "Actual": y_test,
        "Predicted": predictions[best_name]
    })

    comparison = comparison.reset_index(drop=True)

    plt.plot(
        comparison["Actual"].values[:100],
        label="Actual"
    )

    plt.plot(
        comparison["Predicted"].values[:100],
        label="Predicted",
        linestyle="--"
    )

    plt.title("Actual vs Predicted Classes")

else:

    plt.scatter(
        y_test,
        predictions[best_name],
        alpha=0.6
    )

    min_value = min(
        np.min(y_test),
        np.min(predictions[best_name])
    )

    max_value = max(
        np.max(y_test),
        np.max(predictions[best_name])
    )

    plt.plot(
        [min_value, max_value],
        [min_value, max_value],
        linestyle="--"
    )

    plt.xlabel("Actual Values")
    plt.ylabel("Predicted Values")
    plt.title("Actual vs Predicted Values")

plt.legend()
plt.tight_layout()
plt.show()

# ------------------------------------------------
# 18. SAMPLE PREDICTION
# ------------------------------------------------

print("\n" + "="*65)
print("SAMPLE INPUT AND OUTPUT PREDICTION")
print("="*65)

sample_input = X_test.iloc[[0]].copy()

print("\nSAMPLE INPUT VARIABLES:")
display(sample_input)

sample_prediction = best_model.predict(sample_input)

if is_classification:
    predicted_label = encoder.inverse_transform(
        sample_prediction.astype(int)
    )[0]

    print("PREDICTED OUTPUT:", predicted_label)

    print("\nACTUAL OUTPUT:",
          encoder.inverse_transform(
              np.asarray(y_test[:1], dtype=int)
          )[0])

    if hasattr(best_model, "predict_proba"):
        print("PREDICTION PROBABILITIES:")
        display(pd.DataFrame(
            best_model.predict_proba(sample_input),
            columns=encoder.classes_
        ))

else:
    print("PREDICTED OUTPUT:", sample_prediction[0])
    print("ACTUAL OUTPUT:", y_test.iloc[0])

# ------------------------------------------------
# 19. SAVE BEST MODEL AND RESULTS
# ------------------------------------------------

import joblib

joblib.dump(best_model, "Power_Grid_Best_Model.pkl")

results_df.to_csv(
    "Power_Grid_11_Model_Results.csv",
    index=False
)

print("\nFILES SAVED SUCCESSFULLY:")
print("1. Power_Grid_Best_Model.pkl")
print("2. Power_Grid_11_Model_Results.csv")

print("\n" + "="*65)
print("POWER GRID 11 ML MODEL IMPLEMENTATION COMPLETED")
print("="*65)
```

```python

# =====================================================
# EXPLAINABLE AI - POWER GRID PROJECT
# SHAP + LIME
# =====================================================

!pip -q install shap lime

import shap
import lime
import lime.lime_tabular
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.pipeline import Pipeline

print("="*60)
print("EXPLAINABLE AI IMPLEMENTATION")
print("="*60)

# -----------------------------------------------------
# 1. SELECT BEST TRAINED MODEL
# -----------------------------------------------------

print("Selected model:", best_name)

# Transform input data using trained preprocessing pipeline
preprocessor_fitted = best_model.named_steps["preprocessing"]

X_train_transformed = preprocessor_fitted.transform(X_train)
X_test_transformed = preprocessor_fitted.transform(X_test)

feature_names = preprocessor_fitted.get_feature_names_out()

X_train_xai = pd.DataFrame(
    X_train_transformed,
    columns=feature_names
)

X_test_xai = pd.DataFrame(
    X_test_transformed,
    columns=feature_names
)

# Limit samples for faster explanation
background = X_train_xai.sample(
    min(100, len(X_train_xai)),
    random_state=42
)

explain_data = X_test_xai.sample(
    min(50, len(X_test_xai)),
    random_state=42
)

# -----------------------------------------------------
# 2. SHAP EXPLAINER
# -----------------------------------------------------

model = best_model.named_steps["model"]

def predict_function(data):
    data = np.asarray(data)
    return model.predict_proba(data)

explainer = shap.Explainer(
    predict_function,
    background.to_numpy()
)

shap_values = explainer(
    explain_data.to_numpy()
)

print("\nSHAP VALUES GENERATED SUCCESSFULLY")

# -----------------------------------------------------
# 3. SHAP SUMMARY GRAPH
# -----------------------------------------------------

shap.summary_plot(
    shap_values,
    explain_data,
    feature_names=feature_names,
    show=False
)

plt.title("SHAP Summary Plot - Power Grid")
plt.tight_layout()
plt.show()

# -----------------------------------------------------
# 4. SHAP FEATURE IMPORTANCE
# -----------------------------------------------------

values = shap_values.values

# Handle binary and multiclass output
if values.ndim == 3:
    mean_importance = np.abs(values).mean(axis=(0, 2))
else:
    mean_importance = np.abs(values).mean(axis=0)

importance_df = pd.DataFrame({
    "Feature": feature_names,
    "SHAP Importance": mean_importance
}).sort_values(
    "SHAP Importance",
    ascending=False
)

print("\nSHAP FEATURE IMPORTANCE")
display(importance_df.head(15))

plt.figure(figsize=(10, 6))

top_features = importance_df.head(15).sort_values(
    "SHAP Importance"
)

plt.barh(
    top_features["Feature"],
    top_features["SHAP Importance"]
)

plt.xlabel("Mean Absolute SHAP Value")
plt.title("Global SHAP Feature Importance")
plt.tight_layout()
plt.show()

# -----------------------------------------------------
# 5. SHAP WATERFALL PLOT
# -----------------------------------------------------

if values.ndim == 3:
    class_index = 1 if values.shape[2] > 1 else 0
    single_explanation = shap_values[0, :, class_index]
else:
    single_explanation = shap_values[0]

shap.plots.waterfall(
    single_explanation,
    max_display=15
)

# -----------------------------------------------------
# 6. SHAP DEPENDENCE PLOT
# -----------------------------------------------------

most_important_feature = importance_df.iloc[0]["Feature"]

shap.dependence_plot(
    most_important_feature,
    values[:, :, 1] if values.ndim == 3 else values,
    explain_data,
    show=False
)

plt.title("SHAP Dependence Plot")
plt.tight_layout()
plt.show()

# -----------------------------------------------------
# 7. LIME EXPLAINER
# -----------------------------------------------------

lime_explainer = lime.lime_tabular.LimeTabularExplainer(
    training_data=X_train_xai.to_numpy(),
    feature_names=list(feature_names),
    class_names=[str(c) for c in encoder.classes_],
    mode="classification",
    discretize_continuous=True,
    random_state=42
)

# Select one test sample
sample_index = 0

sample = X_test_xai.iloc[sample_index].to_numpy()

lime_explanation = lime_explainer.explain_instance(
    sample,
    predict_function,
    num_features=min(15, len(feature_names)),
    num_samples=3000
)

print("\nLIME EXPLANATION")
print(lime_explanation.as_list())

# -----------------------------------------------------
# 8. LIME VISUALIZATION
# -----------------------------------------------------

fig = lime_explanation.as_pyplot_figure()

plt.title("LIME Local Explanation - Power Grid")
plt.tight_layout()
plt.show()

# -----------------------------------------------------
# 9. LIME EXPLANATION TABLE
# -----------------------------------------------------

lime_df = pd.DataFrame(
    lime_explanation.as_list(),
    columns=["Feature", "Contribution"]
)

display(lime_df)

# -----------------------------------------------------
# 10. ACTUAL VS PREDICTED SAMPLE EXPLANATION
# -----------------------------------------------------

original_sample = X_test.iloc[[sample_index]]

actual_value = y_test[sample_index]
predicted_value = best_model.predict(original_sample)[0]

print("\nSAMPLE EXPLANATION")
print("Actual class:", encoder.inverse_transform(
    np.array([actual_value])
)[0])

print("Predicted class:", encoder.inverse_transform(
    np.array([predicted_value])
)[0])

print("\nEXPLAINABLE AI COMPLETED")

# -----------------------------------------------------
# 11. SAVE XAI RESULTS
# -----------------------------------------------------

importance_df.to_csv(
    "Power_Grid_SHAP_Feature_Importance.csv",
    index=False
)

lime_df.to_csv(
    "Power_Grid_LIME_Explanation.csv",
    index=False
)

print("SHAP and LIME results saved successfully.")
```
